```python
#Preprocessing on a dataset 1
import pandas as pd

data = {
    "Roll no": [101,102,103,104],
    "Name": ["Asha", "Rohan", "Meera", "Karan"],
    "Age": [20, "-",21,"22"],
    "Marks": [85,90,"-",88],
    "Department": ["CSE", "IT", "cse", "ENTC"],
    "Admission_date": ["01-06-2023", "03-06-2023", "05-06-2023", "07-06-2023"]
}

df = pd.DataFrame(data)
#Save as CSV
df.to_csv("student_data.csv", index = False)
```

```python
import pandas as pd
import numpy as np

#Step 1: Load the dataset
df= pd.read_csv("student_data.csv")

print("Original data:")
print(df)
```

```python
df.replace("-", np.nan, inplace = True)
df
```

```python
#Step 2: Replace '-' with NaN to mark the missing values
df.replace("-", np.nan, inplace = True)
#Convert columns to numeric where needed
df["Age"] = pd.to_numeric(df["Age"], errors = 'coerce')
df["Marks"] = pd.to_numeric(df["Marks"], errors = 'coerce')

'''When converting data(like strings to datetime),
if some values are invalid or cannot be converted,
instead of throwing an error, it will:
Replace those invalid values with NaT(Not a Time)'''
df
```

```python
#Step 3: Check missing values
print("\nMissing Values:")
print(df.isnull().sum())
```

```python
#Fill missing marks with median
df["Marks"].fillna(df["Marks"].median(), inplace = True)
df
```

```python
#Fill missing age with number
df["Age"].fillna(df["Age"].mode()[0], inplace = True)
df
```

```python
print(df.isnull().sum())
```

```python
#Step 5: Fix inconsistent text formatting
df["Department"] = df["Department"].str.upper()
df
```

```python
#Step 6: Convert admission date to standard date time format
df["Admission_date"] = pd.to_datetime(df["Admission_date"], errors = 'coerce')
df

'''invalid dates like "abc" or "32-06-2023
```

```python
#Step 7: Show cleaned data
print("\nCleaned Data:")
print(df)
```

```python
#Step 8: Save cleaned dataset
df.to_csv("\nCleaned_student_data.csv", index = False)
print("\nCleaned CSV file saved as cleaned_student_data.csv")
```

```python
#Working with missing values on dataset 2
df = pd.read_csv('/content/Cars93.csv')
df
```

```python
df.isna().sum() #Attributes are displayed
```

```python
#Imputing the missing value
#1. Luggage room
df['Luggage.room']= df['Luggage.room'].fillna(value = df['Luggage.room'].mean())
```

```python
#2. Rear.seat.room
df["Rear.seat.room"] = df['Rear.seat.room'].fillna(value=df['Rear.seat.room'].mean())
```

```python
#Airbags
df['AirBags']=df['AirBags'].fillna(value=df['AirBags'].mode().values[0])
```

```python
#save cleaned dataset
df.to_csv("cleaned_Cars93.csv", index = False)
```
