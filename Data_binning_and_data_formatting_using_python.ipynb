```python
import pandas as pd
import numpy as np
data = {
    'Product': ['Laptop', 'Mobile', 'Tablet', 'Headphones', 'Smartwatch', 'Camera'],
    'Price': [55000, 20000, 30000, 2000, 8000, 45000],
    'Units_sold': [25, 60, 35, 80, 50, 15]
}
df = pd.DataFrame(data)
print("Dataset")
df
```

```python
#Data binning
bins = [0, 10000, 30000, 60000]
labels = ['Low', 'Medium', 'High']
df['Price_Category'] = pd.cut(df['Price'], bins=bins, labels=labels)
df
```

```python
#binning units sold
bins2 = [0, 30, 60, 100]
labels2 = ['Low sales', 'Medium sales', 'High sales']
df['Sales_Category'] = pd.cut(df['Units_sold'], bins=bins2, labels = labels2)
df
```

```python
#Data formating
#Convert data type
print(df.dtypes)
df['Units_sold'] = df['Units_sold'].astype(float)
print(df.dtypes)
```

```python
df
```

```python
#Convert price into float
print(df.dtypes)
df['Price'] = df['Price'].astype(float)
print(df.dtypes)
```

```python
df
```

```python
#Convert text to uppercase
df['Product'] = df['Product'].str.upper()
df
```

```python
#Additional operations
#sort data
sorted_df = df.sort_values(by = 'Price')
sorted_df
```

```python
#Display unique values
print(df['Price_Category'].unique())
```

```python
#Dataset: Food delivery orders
import pandas as pd
import numpy as np
data = {
    'Order_ID': ['01', '02', '03', '04', '05', '06', '07', '08'],
    'Order_Value': [250, 450, 120, 600, 300, 150, 500, 220],
    'Delivery_time': [30, 45, 25, 60, 35, 20, 50, 28],
    'Distance_km': [5, 8, 3, 10, 6, 2, 9, 4]
}
df = pd.DataFrame(data)
df
```

```python
#data binning
bins1 = [0, 20, 40, 60]
labels1 = ['Very Quick', 'Quick', 'Slow']
df['Delivery_speed'] = pd.cut(df['Delivery_speed'], bins=bins1, labels=labels1)
df
```

```python
#Convert distance to float
df['Distance_km'] = df['Distance_km'].astype(float)
df
```

```python
#Convert delivery speed to uppercase
df['Delivery_speed'] = df['Delivery_speed'].str.upper()
df
```

```python
#Sorting
print(df.sort_values(by = 'Order_Value'))
df
```
