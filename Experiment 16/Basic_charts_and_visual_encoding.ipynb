```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
```

```python
data = {
'Days': ['Mon', 'Tue', 'Wed', 'Thu', 'Fri'],
'Study hours': [2, 3, 2.5, 4, 3.5],
'Marks': [50, 60, 55, 70, 65],
'Attendance': [60, 70, 65, 80, 75],
'Sleep hours': [7, 8, 7.5, 9, 8.5],
'Assignments completed': [1, 2, 1, 3, 2],
}
df = pd.DataFrame(data)
df
```

```python
#Line chart
plt.plot(df['Days'], df['Study hours'], marker = 'o')
plt.show()
```

```python
#Line chart(trend)
plt.plot(df['Days'], df['Marks'], marker = 'o')
plt.title("Study hours trend")
plt.xlabel("Days")
plt.ylabel("Study hours")
plt.show()
```

```python
#Advanced line chart
plt.figure(figsize=(7,4))
plt.plot(df['Days'], df['Study hours'], marker = 'o', color = 'green', label = 'Study hours')
plt.plot(df['Days'], df['Marks'], marker = 's', color = 'red', label = 'Marks')
plt.title("Study hours and marks trend")
plt.xlabel("Days")
plt.ylabel("Values")
plt.legend()
```

```python
#Bar chart
plt.bar(df['Days'], df['Marks'], color = 'pink')
plt.title("Marks comparison")
plt.xlabel("Days")
plt.ylabel("Marks")
```

```python
#Compare two values study hours and marks
plt.bar(df['Study hours'], df['Marks'], color = 'green')
plt.title("Study hours vs marks comparison")
plt.xlabel("Study hours")
plt.ylabel("Marks")
plt.show()
```

```python
#Advanced bar graph
bars = plt.bar(df['Days'], df['Marks'], color = 'purple')
for bar in bars:
    y = bar.get_height()
    plt.text(bar.get_x() + bar.get_width()/2, y, str(y),ha = 'center', va = 'bottom')
plt.title("Marks comparison")
plt.xlabel("Days")
plt.ylabel("Marks")
plt.grid(True)
plt.show()
```

```python
#Histogram
#Shows how values are spread
data = {
    'Marks': [50, 60, 55, 70, 65, 75, 80, 85, 90, 45, 68, 72, 77, 82, 88]
}
df = pd.DataFrame(data)
plt.hist(df['Marks'], bins = 6, color = 'pink', edgecolor = 'black', alpha = 0.80)
plt.title("Marks distribution")
plt.xlabel("Marks")
plt.ylabel("Frequency")
plt.show()
```

```python
#Scatter plot
plt.scatter(df['Study hours'], df['Marks'])
plt.title("Study hours vs marks comparison")
plt.xlabel("Study hours")
plt.ylabel("Marks")
plt.show()
```

```python
#Conditional scatter plot
#Example category
df['Result'] = ['Pass', 'Fail', 'Pass', 'Pass', 'Fail']
colors = ['red' if result == 'Fail' else 'green' for result in df['Result']]
plt.scatter(df['Study hours'], df['Marks'], c = colors)
plt.title("Study hours vs marks (result-wise)")
plt.xlabel("Study hours")
plt.ylabel("Marks")
plt.show()
```

#Same graphs using seaborn(better visualization)

```python
#Line plot
sns.lineplot(x = 'Days', y = 'Marks', data = df)
plt.title("Seaborn Line Plot")
plt.show()
```

```python
# Dataset 2 — Sales Performance Data
data2 = {
    'Day':      ['Mon', 'Tue', 'Wed', 'Thu', 'Fri', 'Sat', 'Sun', 'Mon', 'Tue', 'Wed'],
    'Region':   ['North', 'South', 'East', 'West', 'North', 'South', 'East', 'West', 'North', 'South'],
    'Sales':    [200, 220, 250, 270, 300, 350, 400, 280, 260, 290],
    'Profit':   [20,  25,  30,  35,  50,  60,  80,  40,  38,  45],
    'Customers':[30,  35,  40,  45,  50,  65,  80,  55,  48,  52],
    'Category': ['Food', 'Clothing', 'Electronics', 'Food', 'Clothing',
                 'Electronics', 'Food', 'Clothing', 'Electronics', 'Food']
}
df2 = pd.DataFrame(data2)
print(df2)
```

```python
# Dataset 2 — 1. Line Chart (Trend) — Sales Over Days
idx = range(len(df2))
plt.figure(figsize=(9, 4))
plt.plot(idx, df2['Sales'], marker='o', color='blue', linestyle='-.')
plt.xticks(idx, [f"{d}({i+1})" for i, d in enumerate(df2['Day'])], rotation=45, fontsize=8)
plt.title('Sales Over Days (Dataset 2)')
plt.xlabel('Day (entry)')
plt.ylabel('Sales')
plt.tight_layout()
plt.show()
```

```python
# Dataset 2 — Advanced Line Chart — Sales and Profit Trend
idx = range(len(df2))
plt.figure(figsize=(9, 4))
plt.plot(idx, df2['Sales'],   marker='*', label='Sales',  color='blue')
plt.plot(idx, df2['Profit'],  marker='s', label='Profit', color='orange')
plt.xticks(idx, [f"{d}({i+1})" for i, d in enumerate(df2['Day'])], rotation=45, fontsize=8)
plt.title('Sales and Profit Trend (Dataset 2)')
plt.xlabel('Day (entry)')
plt.ylabel('Value')
plt.legend()
plt.tight_layout()
plt.show()
```

```python
# Dataset 2 — 2. Bar Chart — Sales by Day Entry
idx = range(len(df2))
plt.figure(figsize=(9, 4))
plt.bar(idx, df2['Sales'], color='#4169E1', edgecolor='black', linewidth=1.5)
plt.xticks(idx, [f"{d}({i+1})" for i, d in enumerate(df2['Day'])], rotation=45, fontsize=8)
plt.title('Sales Comparison (Dataset 2)')
plt.xlabel('Day (entry)')
plt.ylabel('Sales')
plt.tight_layout()
plt.show()
```

```python
# Dataset 2 — Bar Chart — Customers vs Sales
plt.figure(figsize=(7, 4))
plt.bar(df2['Customers'], df2['Sales'], color='pink', edgecolor='black')
plt.title('Customers vs Sales (Dataset 2)')
plt.xlabel('Customers')
plt.ylabel('Sales')
plt.show()
```

```python
# Dataset 2 — Bar Chart with Value Labels
idx = range(len(df2))
plt.figure(figsize=(9, 4))
bars = plt.bar(idx, df2['Sales'], color='#9932CC')
for bar in bars:
    y = bar.get_height()
    plt.text(bar.get_x() + bar.get_width() / 2, y, str(y), ha='center', va='bottom', fontsize=8)
plt.xticks(idx, [f"{d}({i+1})" for i, d in enumerate(df2['Day'])], rotation=45, fontsize=8)
plt.title('Sales Comparison with Labels (Dataset 2)')
plt.xlabel('Day (entry)')
plt.ylabel('Sales')
plt.grid(axis='y', alpha=0.4)
plt.tight_layout()
plt.show()
```

```python
# Dataset 2 — Grouped Bar Chart — Sales and Profit
import numpy as np
idx = np.arange(len(df2))
width = 0.35
plt.figure(figsize=(10, 4))
plt.bar(idx - width / 2, df2['Sales'],  width, label='Sales',  color='steelblue')
plt.bar(idx + width / 2, df2['Profit'], width, label='Profit', color='coral')
plt.xticks(idx, [f"{d}({i+1})" for i, d in enumerate(df2['Day'])], rotation=45, fontsize=8)
plt.title('Sales vs Profit (Dataset 2)')
plt.xlabel('Day (entry)')
plt.ylabel('Values')
plt.legend()
plt.tight_layout()
plt.show()
```

```python
# Dataset 2 — 3. Histogram — Sales Distribution
mean_sales = np.mean(df2['Sales'])
plt.figure(figsize=(7, 4))
plt.hist(df2['Sales'], bins=4, color='lightblue', edgecolor='black', alpha=0.7)
plt.axvline(mean_sales, color='red', linestyle='--', linewidth=2, label=f'Mean: {mean_sales:.1f}')
plt.title('Sales Distribution (Dataset 2)')
plt.xlabel('Sales')
plt.ylabel('Frequency')
plt.legend()
plt.show()
```

```python
# Dataset 2 — 4. Scatter Plot — Sales vs Profit
plt.figure(figsize=(7, 5))
plt.scatter(df2['Sales'], df2['Profit'], color='blue', s=80)
plt.title('Sales vs Profit (Dataset 2)')
plt.xlabel('Sales')
plt.ylabel('Profit')
plt.show()
```

```python
# Dataset 2 — Conditional Scatter Plot — Category-wise
category_colors = {'Food': 'red', 'Clothing': 'green', 'Electronics': 'blue'}
colors = [category_colors[c] for c in df2['Category']]
plt.figure(figsize=(7, 5))
plt.scatter(df2['Sales'], df2['Profit'], c=colors, s=80)

# Legend
import matplotlib.patches as mpatches
legend_patches = [mpatches.Patch(color=v, label=k) for k, v in category_colors.items()]
plt.legend(handles=legend_patches, title='Category')
plt.title('Sales vs Profit (Category-wise, Dataset 2)')
plt.xlabel('Sales')
plt.ylabel('Profit')
plt.show()
```

```python
# Dataset 2 — Seaborn Line Plot — Sales Over Days
plt.figure(figsize=(9, 4))
sns.lineplot(x=range(len(df2)), y='Sales', data=df2, marker='o', color='navy')
plt.xticks(range(len(df2)), [f"{d}({i+1})" for i, d in enumerate(df2['Day'])], rotation=45, fontsize=8)
plt.title('Seaborn Line Plot — Sales Over Days (Dataset 2)')
plt.xlabel('Day (entry)')
plt.ylabel('Sales')
plt.tight_layout()
plt.show()
```

```python
# Dataset 2 — Seaborn Histogram — Sales Distribution
plt.figure(figsize=(7, 4))
sns.histplot(df2['Sales'], bins=4, color='cornflowerblue', edgecolor='black')
plt.title('Seaborn Histogram — Sales Distribution (Dataset 2)')
plt.xlabel('Sales')
plt.ylabel('Frequency')
plt.show()

```

```python
# Dataset 2 — Seaborn Scatter Plot — Sales vs Profit
plt.figure(figsize=(7, 5))
sns.scatterplot(x='Sales', y='Profit', hue='Category', data=df2, s=100)
plt.title('Seaborn Scatter Plot — Sales vs Profit by Category (Dataset 2)')
plt.xlabel('Sales')
plt.ylabel('Profit')
plt.show()

```
